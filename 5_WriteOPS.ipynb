- default mode - errorifexists
- overwrite - it will remove the exisiting data overwrite the new data
- ignore - do nothing, simply come out
- append - add to the existing data

## write into Table

```python
table_schema="BranchID integer,EmployeeID integer,FirstName string,Hiredate string,LastName string,Position string"
table_df=spark.read.csv("/Volumes/izwd37dev/wd37db/rawdatta/csv/emp.csv",header=True,schema=table_schema)
table_df.write.saveAsTable("izwd37dev.wd37db.Emp2_table")
```

```python
table_df=spark.sql("select * from izwd37dev.wd37db.Emp2_table")
table_df.count()
table_df.show()
```


# parquet 

**parquet** - columnar serilized , intelligent , self describing file format

all parquet data stored in binary, columnar format, not as plain text. 




## show() and display() difference

both show() and display() are used to view DataFrames

- show() prints the first 20 rows by default and is part of the Spark DataFrame API
- display() creates an interactive table where you can: Sort columns, Filter data etc.


## Write into parquet file

```python

# write dataframe into serilized file format (parquet ) -- format to byte format
table_df.write.mode("overwrite").parquet("/Volumes/izwd37dev/wd37db/rawdatta/parquet/parquet_table_write")
parquet_df=spark.read.parquet("/Volumes/izwd37dev/wd37db/rawdatta/parquet/parquet_table_write")
parquet_df.show()
parquet_df.count()
display(parquet_df)
```

cannot view parquet file data using **%fs head** command, because all parquet data stored in binary, columnar format, not as plain text.

# Create dataframe using python


Create dataframe - programmatically using python
list(tuple) - [(),(),()]

```python
data=[(100,"laptop",150000,10),(101,"mobile",150000,15),(102,"tv",200000,5),(103,"fridge",14323,10)]
df=spark.createDataFrame(data,["id","name","price","quantity"])
df.show()
print(df.count())
df.write.mode("overwrite").saveAsTable("izwd37dev.wd37db.inventory")


```
